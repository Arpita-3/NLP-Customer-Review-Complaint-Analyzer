```python
from google_play_scraper import reviews, Sort
import pandas as pd
```

```python
APP_ID = "in.swiggy.android"
REVIEW_COUNT = 5000
OUTPUT_FILE = r"F:\pojects for github\NLP-Based Customer Review & Complaint Analyzer\review-analyzer\data\raw\swiggy_reviews_ipynb.csv"
```

```python
result, _ = reviews(
    APP_ID,
    lang="en",
    country="in",
    sort=Sort.NEWEST,
    count=REVIEW_COUNT
)
```

```python
df = pd.DataFrame(result)
```

```python
df.to_csv(OUTPUT_FILE, index=False)
```

```python
print(f"Reviews collected: {len(df)}")
print(f"Reviews saved to: {OUTPUT_FILE}")
```
